```python
from transformers import AutoTokenizer,AutoModel
import torch
```

```python
model_name="bert-base-uncased"
```

```python
print(type(model_name))
```

```python
text="I am waheed Ullah ,currently studying in the fast nuces peshawar campus"
```

```python
tokenize=AutoTokenizer.from_pretrained(model_name)
model=AutoModel.from_pretrained(model_name)
```

```python
token=tokenize.tokenize(text)
```

```python
token
```

```python
encoded=tokenize(text)
```

```python
print(encoded)
```

```python
ids=tokenize.convert_tokens_to_ids(token)
```

```python
ids
```

```python
sentence=tokenize.convert_ids_to_tokens(ids)
```

```python
ids.shape
```

```python
sentence
```

```python
ids=torch.tensor(ids,dtype=torch.long)
```

```python
print(ids.shape)
```

```python
ids
```

```python
print(ids.shape)
```

```python
embedding=model.embeddings.word_embeddings(ids)
```
