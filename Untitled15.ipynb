```python
import torch
import torch.nn as nn
```

```python
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python


class MyRNNCELL(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        # Initialize weights and biases as trainable parameters
        self.weight_ih = nn.Parameter(torch.randn(hidden_size, input_size))
        self.weight_hh = nn.Parameter(torch.randn(hidden_size, hidden_size))
        self.bias_ih = nn.Parameter(torch.randn(hidden_size))
        self.bias_hh = nn.Parameter(torch.randn(hidden_size))

    def forward(self, input, hidden):
        # Linear transformation: W_ih @ x + W_hh @ h + b_ih + b_hh
        h1 = (torch.matmul(input, self.weight_ih.T) +
              torch.matmul(hidden, self.weight_hh.T) +
              self.bias_ih + self.bias_hh)
        h1_out = torch.tanh(h1)
        return h1_out


class MYRNN(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        self.hidden_size = hidden_size
        self.myrnn = MyRNNCELL(input_size, hidden_size)

    def forward(self, x):
        batch_size = x.shape[0]
        seq_length = x.shape[1]

        # Initialize hidden state with zeros on the same device as input
        hidden = torch.zeros(batch_size, self.hidden_size, device=x.device)
        output = []

        for i in range(seq_length):
            word = x[:, i, :]
            # Update hidden state sequentially
            hidden = self.myrnn(word, hidden)
            output.append(hidden)

        # Stack outputs along sequence dimension (batch_size, seq_length, hidden_size)
        output = torch.stack(output, dim=1)

        return output, hidden
```

```python
x = torch.randn(2, 3, 4)
model = MYRNN(input_size=4, hidden_size=5)
output, hidden = model(x)

print("Output shape:", output.shape)  # torch.Size([2, 3, 5])
print("Hidden shape:", hidden.shape)  # torch.Size([2, 5])
```
